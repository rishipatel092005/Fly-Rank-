## My Lane as an ML Task

My lane is **Ranking**.

The goal is to rank content pages based on their refresh priority. Instead of reviewing every page manually, the model orders pages from highest to lowest priority using multiple SEO signals. This helps the content team decide which pages should be refreshed first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The target is the **refresh priority rank** of each content page.

Since the dataset does not contain a direct "refresh priority" label, the ranking is treated as a proxy based on observed SEO signals such as CTR, impressions, trend direction, search position, and other available features. The model uses these observed values to estimate which pages should receive higher priority for review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

A good model is one that places the highest-priority pages near the top of the ranked list.

Success is measured by how well the predicted ranking matches the expected priority of pages. A better ranking means the content team reviews the most important pages first, improving decision support.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A simple if-statement cannot capture the complex relationship between multiple SEO signals.

For example, a page may have a high CTR but declining impressions, while another page may have lower CTR but rapidly increasing search visibility. Machine learning can learn these patterns from the data and produce a better ranking than a fixed rule based on only one or two conditions.

The model supports decision-making by considering many features together instead of relying on manually defined thresholds.
